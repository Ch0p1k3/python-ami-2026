# Генераторы

**Генератор**

Это итератор, который получается в результате вызова функции, содержащей ключевое слово `yield`.

```python
import typing as tp

def countdown(n: int) -> tp.Iterator[int]:
    print(f'Counting down from {n}')
    for i in range(n, 0, -1):
        yield i
    print('Done')
```

```python
counter = countdown(2)
counter
```

```python
iter(counter) is counter
```

```python
next(counter)
```

```python
next(counter)
```

```python
next(counter)
```

```python
def magic_countdown(n: int) -> tp.Iterator[int]:
    yield 42
    yield from countdown(n)
```

```python
counter = magic_countdown(5)
counter
```

```python
for n in counter:
    print(n)
```

##  islice, count, cycle, repeat

```python
from itertools import count, cycle, repeat, islice
```

```python
sequence = range(10)
```

```python
list(islice(sequence, 2, 5))  # seq[2:5]
```

```python
list(islice(sequence, 2, 8, 2))  # seq[2:5]
```

```python
list(islice(count(start=0, step=3), 10))
```

```python
list(islice(cycle('kek'), 9))
```

```python
list(islice(repeat(42), 4))
```

## product

```python
roles = ("bs", "yabs", "bsrank")
modes = ("meta", "stat")

for r in roles:
    for m in modes:
        print(f"{r=}, {m=}")
```

## Deque

```python
from collections import deque

s = ['ab', 'ab', 'bc', 'cd', 'ab']
d = deque(s)
print(d)

print('-' * 30)

for elem in d:
    print(elem)

print('-' * 30)

# Добавление элементов

d.append('aa')
d.appendleft('dd')
d.extend(['hh', 'pp'])
d.extendleft(['hh', 'pp'])
print(d)

print('-' * 30)

# Удаление элементов

print(d.pop())
print(d.popleft())
print(d)

print('-' * 30)

# Обращение к элементам

print(d[0])
print(d[-2])
print(d.count('hh'))

print('-' * 30)

# Смещение очереди влево-вправо

d.rotate(1) # влево
# 1-2-3-4-5 -> rotate(1) -> 2-3-4-5-1
print(d)
d.rotate(-2) # вправо
# 1-2-3-4-5 -> rotate(-2) -> 4-5-1-2-3
print(d)
d.reverse() # перевернуть очередь
print(d)

print('-' * 30)

# вставка и удаление

d.insert(1, 'x')
print(d)
print(d.index('x'))
d.remove('x')
print(d)
```

## heapq

```python
import heapq

a = []
print(a)
heapq.heappush(a, 1)
print(a)
heapq.heappush(a, 2)
print(a)
heapq.heappush(a, 0)
print(a)
heapq.heappush(a, 4)
print(a)
heapq.heappush(a, -1)
print(a)
```

```python
while a:
    value = heapq.heappop(a)
    print(value, a)
```

```python
 # Получить 3 максимальных элемента

a = []
for value in [1, 5, 7, 2, 4, 3]:
    print(a)
    if len(a) != 3:
        heapq.heappush(a, value)
    elif a[0] < value:
        heapq.heapreplace(a, value)
a
```
